This notebook is used to count all the authors of accepted papers.
Be aware that authors of multiple papers could be counted several times if names are written slightly differently.

Michele Scarpiniti -- DIET Dpt. (Sapienza University of Rome)

```python
import IJCNN_utils as ut
```

```python
# Set file names
path = './data/'
paper_file = path + 'All_Accepted_Papers.xlsx'

# Load paper data
papers = ut.load_submissions(paper_file)
```

```python
# Main loop
L = []
for i in range(len(papers)):
    a = ut.getAuthorList2(papers.iloc[i]['Author Names'])

    b = [s.strip() for s in a]  # Remove spaces
    L = L + b    # COncat to previous authors
```

```python
# Extract unique entries and count
L_unique = list(set(L))
N_auth = len(L_unique)

print('Number of accepted authors =', N_auth)
```
